**Ejercicio 3:**

He hecho dos MLP, uno de 3 capas ocultas de tamaños [128, 64, 32] y uno de una sola capa de [256]. Para alfa 0.5 y 2000 iteraciones ambos han tenido estos resultados.

El primero (3 capas):
MLP Accuracy for Lambda = 0.50000 : 0.88298
MLP Confusion Matrix:
[[ 1  2  5]
[ 3 45  0]
[ 0  1 37]]

El segundo (1 capa):
MLP Accuracy for Lambda = 0.50000 : 0.90426
MLP Confusion Matrix:
[[ 1  2  5]
[ 1 47  0]
[ 0  1 37]]


 El formato de las matrices de confusión de SKLearn es este:
 x -> clases predichas
 y -> clases reales
 Por tanto, y asumiendo que los valores están ordenados en orden alfabético (Converted, Demented, Nondemented)

|             | Pred Convr. | Pred. Demen. | Pred. Nond |
| ----------- | ----------- | ----------- | ----------- | 
| Real Convr. | 1 vs 1 | 2 vs 2 | 5 vs 5 |
| Real Demen. | 3 vs 1 | 45 vs 47 | 0 vs 0 |
| Real Nonde. | 0 vs 0 | 1 vs 1 | 37 vs 37 |

 En general ambos han tenido buena accuracy, pero lo más peligroso en algo como esto son los falsos negativos, pudiendo pasar por desapercibida la enfermedad en un paciente.

 El primero ha tenido algo más de cofusión entre Dementes y Conversos, entendible siendo Conversos un valor tan limitante entre Demente y No demente, pero el segundo no tanto. Ambos peligran mucho en que han tenido 5 casos de Conversos reales que han declarado como No Dementes, falsos negativos. 

 En general escogería el segundo con una sola capa porque ha tenido más precisión general y es más simple pero ambos tienen el mismo potencial defecto. Bien es cierto que la cantidad de datos es limitada y habría que verlos con más.
